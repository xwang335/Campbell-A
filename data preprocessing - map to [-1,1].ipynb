```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('/Users/ywo/Downloads/Industry Project/feature construction.csv')
```

```python
df = df[df['ret'].notna()]
df = df.reset_index(drop=True)
```

normalization to [-1,1]

```python
def norm_rank(data):
    ranks=data.rank(method='average',na_option='keep')
    n=ranks.count()
    mapped = (ranks / (n + 1)) * 2 - 1
    return mapped.fillna(0)
```

```python
features=['beta','betasq','mvel1','retvol','chmom']
```

```python
df[features]=df.groupby('date')[features].apply(norm_rank).reset_index(drop=True)
print(df)
```
